## Tema Analisis
**Analisis Karakteristik Properti Berdasarkan Ukuran Bangunan dan Fasilitas Rumah**

## Tujuan Analisis
- Mengidentifikasi karakteristik rumah berdasarkan jumlah kamar dan luas bangunan.
- Menganalisis hubungan fasilitas rumah dengan ukuran properti.
- Mengetahui pola kepemilikan properti berdasarkan data yang tersedia.
- Membandingkan variasi luas bangunan pada setiap properti.
- Memberikan informasi sederhana untuk memahami karakteristik aset properti.


```python
# Import Library
import pandas as pd
import matplotlib.pyplot as plt

# Membaca dataset
df = pd.read_csv('property_without_duplicates.csv')

df.head()
```

## Informasi Dataset

Dataset berisi informasi properti seperti:
- Nomor properti
- Nama jalan
- Status kepemilikan
- Jumlah kamar tidur
- Jumlah kamar mandi
- Luas bangunan


```python
# Jumlah data dan kolom
print("Jumlah Data:", df.shape[0])
print("Jumlah Kolom:", df.shape[1])

pd.DataFrame({
    'No': range(1, len(df.columns)+1),
    'Nama Kolom': df.columns
})
```

## Analisis 1: Perbandingan Luas Bangunan

Analisis ini digunakan untuk melihat variasi ukuran rumah berdasarkan luas bangunan.


```python
plt.figure(figsize=(8,5))

plt.bar(
    df['ST_NAME'],
    df['SQ_FT']
)

plt.title('Perbandingan Luas Bangunan Setiap Properti')
plt.xlabel('Lokasi Properti')
plt.ylabel('Luas Bangunan (SQ FT)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## Analisis 2: Distribusi Jumlah Kamar Tidur

Analisis ini bertujuan melihat tipe rumah berdasarkan jumlah kamar tidur.


```python
bedroom = df['NUM_BEDROOMS'].value_counts().sort_index()

plt.figure(figsize=(7,4))

plt.bar(
    bedroom.index.astype(str),
    bedroom.values
)

plt.title('Distribusi Jumlah Kamar Tidur')
plt.xlabel('Jumlah Kamar Tidur')
plt.ylabel('Jumlah Properti')

plt.tight_layout()
plt.show()
```

## Analisis 3: Status Kepemilikan Rumah

Analisis ini melihat perbandingan rumah yang dihuni pemilik dan bukan pemilik.


```python
ownership = df['OWN_OCCUPIED'].value_counts()

plt.figure(figsize=(7,4))

plt.bar(
    ownership.index,
    ownership.values
)

plt.title('Status Kepemilikan Properti')
plt.xlabel('Status Kepemilikan')
plt.ylabel('Jumlah Properti')

plt.tight_layout()
plt.show()
```

# Kesimpulan Analisis

Berdasarkan hasil analisis dataset properti, dapat disimpulkan bahwa:

1. Dataset menunjukkan adanya variasi karakteristik rumah berdasarkan luas bangunan, jumlah kamar tidur, jumlah kamar mandi, dan status kepemilikan.

2. Perbedaan luas bangunan menunjukkan bahwa setiap properti memiliki ukuran yang berbeda sehingga dapat menggambarkan variasi tipe hunian dalam dataset.

3. Berdasarkan jumlah kamar tidur, sebagian besar properti berada pada tipe rumah dengan jumlah kamar yang relatif sederhana, sehingga menunjukkan karakteristik hunian yang lebih banyak digunakan untuk kebutuhan keluarga kecil hingga menengah.

4. Status kepemilikan menunjukkan adanya perbedaan antara properti yang dihuni oleh pemilik dan properti yang kemungkinan digunakan sebagai investasi atau disewakan.

5. Secara keseluruhan, analisis data properti dapat membantu memahami pola aset berdasarkan ukuran, fasilitas, dan kepemilikan sehingga informasi tersebut dapat digunakan sebagai dasar evaluasi karakteristik rumah.
